```python
from PIL import Image
import numpy as np
from google.colab import files
import io
```

```python
uploaded = files.upload()
image = Image.open(io.BytesIO(list(uploaded.values())[0])).convert("RGB")
```

```python
img_array = np.array(image)
r, g, b = np.mean(img_array, axis=(0, 1)).astype(int)
```

```python
known_colors = {
    "Black": (0, 0, 0),
    "White": (255, 255, 255),
    "Red": (255, 0, 0),
    "Lime": (0, 255, 0),
    "Blue": (0, 0, 255),
    "Yellow": (255, 255, 0),
    "Cyan": (0, 255, 255),
    "Magenta": (255, 0, 255),
    "Gray": (128, 128, 128),
    "Maroon": (128, 0, 0),
    "Olive": (128, 128, 0),
    "Green": (0, 128, 0),
    "Purple": (128, 0, 128),
    "Teal": (0, 128, 128),
    "Navy": (0, 0, 128),
    "Orange": (255, 165, 0),
    "Pink": (255, 192, 203),
    "Brown": (165, 42, 42),
    "Gold": (255, 215, 0),
    "Sky Blue": (135, 206, 235)
}
def defineColor(r,g,b):
  dif=float("inf")
  x=0
  chosenColor=None
  for color,rgb in known_colors.items():
    x=abs(r-rgb[0])+abs(g-rgb[1])+abs(b-rgb[2])
    if x<dif:
      chosenColor=color
      dif=x
  return chosenColor
```

```python
color=defineColor(r,g,b)
print(color)
```
